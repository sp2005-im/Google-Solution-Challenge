```python
#imports
import pandas as pd
from chembl_webresource_client.new_client import new_client
```

```python
#Search for Target Protein
#Target search for coronavirus
target = new_client.target
target_query = target.search('coronavirus')
targets = pd.DataFrame.from_dict(target_query)
targets
```

```python
#select and retrieve biological activity for SARS coronavirus 3C-like proteinase
selected_target = targets.target_chembl_id[6]
selected_target
```

```python
activity = new_client.activity
res = activity.filter(target_chembl_id = selected_target).filter(standard_type="IC50")
```

```python
df = pd.DataFrame.from_dict(res)
```

```python
df.head(5)
```

```python
df.standard_type.unique()
```

```python
#Saving the bioactivity data to a CSV file -> bioactivity_data.csv
df.to_csv('bioactivity_data.csv', index = "False") #- We don't want the resulting index number in the csv file
```

```python
l = df.standard_value
p = 0
for i in l:
  if p == None:
    p+=1
print(p)
```

```python
df2 = df[df.standard_value.notna()]
df2.head(5)
```

```python
l1 = df2.standard_value
print(len(l1))
t = 0
for i in l1:
  if i == None:
    t+=1
print(t)
```

```python
#labelling compounds as active, inactive or intermediate based on IC 50 value
bioactivity_class = []
for i in df2.standard_value:
  if float(i)>=10000:
    bioactivity_class.append("inactive")
  elif float(i)<=1000:
    bioactivity_class.append("active")
  else:
    bioactivity_class.append("intermediate")
```

```python
print(df2.molecule_chembl_id) #It won;t be 247 because we removed the null values
```

**The above result shows 245 entries not 247 entries as two None values were removed**

```python
mol_cid = []
for i in df2.molecule_chembl_id:
  mol_cid.append(i)
```

```python
len(mol_cid)
```

```python
df2.canonical_smiles
```

```python
#Cannonical Smiles - Simplified Molecular Input Line Entry System (SMILES) - Although unique for each structure, dependent on the cannonicalization
# Algorithm used
canonical_smiles = []
for i in df2.canonical_smiles:
  canonical_smiles.append(i)
```

```python
len(canonical_smiles)
```

```python
standard_value = []
for i in df2.standard_value:
  standard_value.append(i)
len(standard_value)
```

```python
data_tuples = list(zip(mol_cid, canonical_smiles, bioactivity_class, standard_value))
df3 = pd.DataFrame(data_tuples, columns = ['molecule_chembl_id', 'canonical_smiles', 'bioactivity_class', 'standard_value'])
```
